```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
file_path = '/content/drive/MyDrive/Colab Notebooks/Assignment -2/NBS Data Sheet.csv'
df = pd.read_csv(file_path)
print("Original Dataset:")
print(df)
```

```python
le = LabelEncoder()
df_encoded = pd.DataFrame()
```

```python
for col in df.columns:
    df_encoded[col] = le.fit_transform(df[col])
```

```python
X = df_encoded.drop('Heart Disease', axis=1)
y = df_encoded['Heart Disease']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=54)
```

```python
model = CategoricalNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("-" * 30)
```

```python
sample_input = pd.DataFrame([X.iloc[0]]) # Example: using the first row of X as a sample input
prediction = model.predict(sample_input)
print(f"Prediction for sample input: {prediction[0]}")
```

```python
result = "Yes" if prediction[0] == 1 else "No"
print(f"Prediction for Sample Input (Heart Disease): {result}")
```
